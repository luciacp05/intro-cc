```python
Algoritmo seguimiento2a
	Definir n, d Como Entero
	Definir esPrimo Como Logico
	Escribir "Ingrese n:"
	Leer n
	d <- 2
	esPrimo <- (n >= 2)
	Mientras d * d <= n Y esPrimo Hacer
		Si n mod d = 0 Entonces
			esPrimo <- Falso
		FinSi
		Escribir d, " | ", n mod d = 0, " | ", esPrimo
		d <- d + 1
	FinMientras
	Escribir "esPrimo = ", esPrimo
FinAlgoritmo
```

```python
n = int(input("Ingrese n: "))
d = 2
esPrimo = (n >= 2)
while d * d <= n and esPrimo:
    es_divisor = (n % d == 0)
    if es_divisor:
        esPrimo = False
    print(f"{d} | {str(es_divisor).upper()} | {str(esPrimo).upper()}")
    d += 1
print(f"esPrimo = {esPrimo}")
```
